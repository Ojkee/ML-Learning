```python
from config import ModelConfig

model_cfg = ModelConfig(
    name="BaseFromPaper",
    d_vocab=32_000,
    context_len=128,
    d_k=64,
    d_v=64,
    d_model=64,
    h=8,
    N=6,
    dropout_rate=0.1,
)
```

```python
from tokenizer import TokenizerBPE

tokenizer = TokenizerBPE(name=f"{model_cfg.d_vocab//1_000}k_shared", vocab_size=model_cfg.d_vocab)
```

```python
from numpy import average, median, percentile


def tokenized_stats(file):
    lens = [len(tokenizer.encode(line)) for line in file.readlines()]
    return {
        "min: ": min(lens),
        "p25: ": percentile(lens, 25),
        "median: ": median(lens),
        "p75: ": percentile(lens, 75),
        "max: ": max(lens),
        "avg: ": average(lens),
    }

with open("./opus100/opus.en-pl-dev.en") as en_dev_file, \
    open("./opus100/opus.en-pl-train.en") as en_train_file, \
    open("./opus100/opus.en-pl-test.en") as en_test_file, \
    open("./opus100/opus.en-pl-dev.pl") as pl_dev_file, \
    open("./opus100/opus.en-pl-train.pl") as pl_train_file, \
    open("./opus100/opus.en-pl-test.pl") as pl_test_file:

    print(f"{tokenized_stats(en_dev_file)=}")
    print(f"{tokenized_stats(en_train_file)=}")
    print(f"{tokenized_stats(en_test_file)=}")
    print(f"{tokenized_stats(pl_dev_file)=}")
    print(f"{tokenized_stats(pl_train_file)=}")
    print(f"{tokenized_stats(pl_test_file)=}")
```
